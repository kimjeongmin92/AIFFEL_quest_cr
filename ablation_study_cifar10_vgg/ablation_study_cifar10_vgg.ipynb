이번 실습에서는 CIFAR-10 데이터셋을 사용하여 VGGNet을 구현하고, VGG-16과 VGG-19의 구조 및 성능을 비교한다.  
또한 Ablation Study의 개념을 바탕으로 모델 구조의 차이가 성능에 어떤 영향을 주는지 확인한다.

````python
# Q. Dense Block 내의 각 레이어 output의 channel을 계산하는 함수를 적어보세요.

def get_channel_list():
    channel_list = []
    input_channel = 32
    growth_rate = 32

    for i in range(12):
        channel_list.append(input_channel + growth_rate * i)

    return channel_list

get_channel_list()
````

# Ablation Study 실습: CIFAR-10 기반 VGGNet 비교

## 1. 학습 목표

이번 실습의 목표는 Ablation Study의 개념을 이해하고, CIFAR-10 데이터셋을 활용하여 VGGNet을 직접 구현해보는 것이다.  
또한 VGG-16과 VGG-19를 비교하여 모델 구조의 차이가 성능에 어떤 영향을 주는지 확인한다.

## 2. Ablation Study란?

Ablation Study는 모델의 특정 기법이나 구성 요소가 실제로 성능 향상에 기여했는지 확인하기 위한 실험이다.  
여러 방법이 함께 적용된 모델에서 일부 요소를 제거한 경우와 적용한 경우를 비교함으로써, 각 요소의 효과를 객관적으로 검증할 수 있다.

## 3. CIFAR-10 데이터셋

CIFAR-10은 32×32 크기의 컬러 이미지 6만 장으로 구성된 이미지 분류 데이터셋이다.  
총 10개의 클래스를 가지고 있으며, 이번 실습에서는 이 데이터셋을 사용하여 VGGNet 모델을 학습하고 비교한다.

## 4. Dense Block 채널 계산 실습

DenseNet에서는 각 레이어가 이전 레이어들의 feature map을 모두 입력으로 받는다.  
따라서 레이어가 깊어질수록 입력 channel 수가 growth rate만큼 증가한다.

이번 실습에서는 초기 입력 채널을 32, growth rate를 32, Dense Block 내 레이어 수를 12로 설정하여 각 레이어의 channel 수를 계산하였다.

```python
def get_channel_list():
    channel_list = []
    input_channel = 32
    growth_rate = 32

    for i in range(12):
        channel_list.append(input_channel + growth_rate * i)

    return channel_list

get_channel_list()
